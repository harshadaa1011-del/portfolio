```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure
%matplotlib inline
matplotlib.rcParams['figure.figsize']=(12,8)

df=pd.read_csv(r"C:\Users\sanji\Downloads\archive (8).zip")
df.head()
```

```python
df= df.fillna(0)
```

```python
df.dtypes
```

```python
df['budget']=df['budget'].astype('int64')
df['gross']=df['gross'].astype('int64')
df['runtime']=df['runtime'].astype('int64')
df['votes']=df['votes'].astype('int64')

df
```

```python
df["correct_year"]=df["released"].str.extract(r"(\d{4})")
df
```

```python
df=df.sort_values(by=["gross"],ascending=False)
df
```

```python
df=df.drop_duplicates()
df.head()
```

```python
plt.scatter(x=df["budget"],y=df["gross"])
plt.title("budget vs gross earnings ")
plt.xlabel("gross earning  ")
plt.ylabel("budget for film")
```

```python
sns.regplot(x='budget',y='gross',data=df,scatter_kws={'color':'red'},line_kws={'color':'blue'})
```

```python
df.corr(numeric_only=True)
```

```python
correlation_matrix=df.corr(numeric_only=True , method="kendall")
sns.heatmap(correlation_matrix,annot=True)
plt.title('correlation matric for numeric features')
plt.xlabel('movie features')
plt.ylabel('movie features')

plt.show()
```

```python
df_numerized = df.copy()
for col_name in df_numerized.columns:
        df_numerized[col_name]=df_numerized[col_name].astype('category')
        df_numerized[col_name]=df_numerized[col_name].cat.codes


df_numerized

```

```python

correlation_matrix=df_numerized.corr(numeric_only=True , method="kendall")
sns.heatmap(correlation_matrix,annot=True)
plt.title('correlation matric for numeric features')
plt.xlabel('movie features')
plt.ylabel('movie features')

plt.show()
```

```python
correlation_mat=df_numerized.corr()
corr_pairs = correlation_mat.unstack()
corr_pairs
```

```python
sorted_pairs= corr_pairs.sort_values()
sorted_pairs
```

```python
high_corr = sorted_pairs[(sorted_pairs)>0.5]
high_corr
```
