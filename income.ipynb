```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure
%matplotlib inline
matplotlib.rcParams['figure.figsize']=(12,8)
```

```python
df=pd.read_csv(r"C:\Users\sanji\Downloads\archive (10).zip")
df
```

```python
df.drop_duplicates()
```

```python
df=df.fillna("")
df
```

```python
df=df.replace("?","")
df.head()
```

```python
df=df.sort_values(by=["income","capital.gain","capital.loss","hours.per.week"],ascending=[False,False,True,True])
df
```

```python
plt.figure(figsize=(8,4))
sns.barplot(y="capital.gain",x="sex",data=df )
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(x="sex",hue="income",data=df)

plt.show()
```

```python
df[df['income'] == '>50K']['sex'].value_counts()
```

```python
df_numerized=df
for col_name in df_numerized.columns:
    df_numerized[col_name]=df_numerized[col_name].astype("category")
    df_numerized[col_name]=df_numerized[col_name].cat.codes

df_numerized
```

```python
correlation_matrix=df_numerized.corr(numeric_only=True,method="pearson")
sns.heatmap(correlation_matrix,annot=True)
plt.title("correlation matrix")
plt.show()
```
